# Part A — Week 7 Review
## 1. Modules

A **module** is a Python file that contains reusable code.

A module can contain:

- functions,
- variables,
- classes,
- and other Python statements.

Modules help keep larger programs organized and allow code to be reused instead of rewritten.


### Example: Using a Standard-Library Module

The `math` module contains mathematical functions and constants.

After importing the module, use the module name followed by a dot to access its contents.


````python
import math

print(math.sqrt(25))
print(math.pi)
````

## 2. Common Import Styles

Python supports several common import styles.

### Import the entire module

```python
import math
```

Use the module name when calling a function:

```python
math.sqrt(16)
```

### Import with an alias

```python
import math as m
```

Now `m` is a shorter name for the module:

```python
m.sqrt(16)
```

### Import a specific item

```python
from math import sqrt
```

Now you can call the function directly:

```python
sqrt(16)
```


### Example: Compare Import Styles

Run the cell and notice that all three approaches use functions from the same module.


````python
import math
print(math.ceil(4.2))

import math as m
print(m.floor(4.8))

from math import sqrt
print(sqrt(81))
````

## 3. The `datetime` Module

Python's `datetime` module provides tools for working with dates and times.

Important objects include:

- `date` — represents a calendar date,
- `datetime` — represents both date and time,
- `timedelta` — represents a duration or difference between dates or times.


### Example: Working with Dates

Subtracting one `date` object from another produces a `timedelta`.
The `.days` attribute gives the number of days in that difference.


````python
from datetime import date

start = date(2026, 9, 1)
end = date(2026, 9, 22)

difference = end - start

print(difference)
print("Number of days:", difference.days)
````

## 4. Moving Forward in Time with `timedelta`

A `timedelta` can also be added to a date.

For example, `timedelta(days=7)` represents one week.


````python
from datetime import date, timedelta

today = date(2026, 9, 22)
one_week = timedelta(days=7)

next_week = today + one_week

print("Today:", today)
print("One week later:", next_week)
````

## 5. Module vs. Package vs. Library

These terms are related, but they are not exactly the same.

- **Module:** usually one `.py` file containing reusable code.
- **Package:** a collection of related Python modules organized together.
- **Library:** a broader collection of reusable tools that may include modules and packages.

A simple way to remember the relationship:

**module → one file**  
**package → related modules**  
**library → broader reusable toolkit**


## 6. Virtual Environments

A **virtual environment** is an isolated Python environment for a specific project.

Different projects may require different versions of the same package.

A virtual environment helps prevent dependencies from one project from interfering with another project.

A typical workflow begins with:

```bash
python -m venv myenv
```

Packages can then be installed inside the environment using `pip`.


## 7. Installing vs. Importing

Installing and importing are different steps.

### Installing

```bash
pip install pandas
```

This makes the package available in the Python environment.

### Importing

```python
import pandas
```

This makes the installed package available inside the Python program.

**install → environment**  
**import → program**


## 8. Reusing Code from Another Python File

Reusable functions can be stored in one `.py` file and imported into another.

Suppose `my_utils.py` contains:

```python
def double(number):
    return number * 2
```

Another file can use that function with:

```python
from my_utils import double
```

This is one of the main ideas behind **modular programming**: organize a larger program into smaller reusable pieces.

## 9. What Is Object-Oriented Programming?

**Object-Oriented Programming (OOP)** is a way to organize a program by grouping related:

- **data**, and
- **actions that work with that data**

together.

OOP can help keep larger programs organized, reusable, and easier to maintain.


## 10. Class vs. Object

A **class** is a blueprint or template.

An **object** is one actual instance created from that class.

A simple analogy:

- class → cookie cutter
- object → cookie made from the cutter

One class can be used to create many objects.


### Example: Class and Objects

The class defines the structure once.  
Each object can store different values.


````python
class Cookie:
    def __init__(self, shape, flavor):
        self.shape = shape
        self.flavor = flavor

cookie1 = Cookie("star", "chocolate")
cookie2 = Cookie("heart", "vanilla")

print(cookie1.shape, cookie1.flavor)
print(cookie2.shape, cookie2.flavor)
````

## 11. Attributes and Methods

An **attribute** is data stored inside an object.

A **method** is a function that belongs to a class.

For example, a student object might have:

### Attributes
- `name`
- `major`
- `credits`

### Methods
- `add_credits()`
- `summary()`

A useful rule:

**attribute → data**  
**method → behavior/action**


## 12. `__init__()` and `self`

`__init__()` is a special method that Python calls automatically when a new object is created.

It is commonly used to set an object's starting attributes.

`self` refers to the **current object**.

For example:

```python
self.name = name
```

stores the value of `name` inside the current object.


### Example: Student Class

Run the cell and notice how each object gets its own data.


````python
class Student:
    def __init__(self, name, major, credits=0):
        self.name = name
        self.major = major
        self.credits = credits

    def summary(self):
        return f"{self.name} - {self.major} - {self.credits} credits"

student1 = Student("Sarah", "Data Science", 9)
student2 = Student("David", "Computer Science")

print(student1.summary())
print(student2.summary())
````

## 13. Dot Notation

A dot (`.`) is used to access an object's attributes and methods.

### Attribute

```python
student1.name
```

### Method

```python
student1.summary()
```

A simple way to recognize them:

- attributes usually do not use parentheses,
- methods are called with parentheses.


## 14. Inheritance and Subclasses

A **subclass** is a more specific class that inherits from another class.

Inheritance lets a child class reuse attributes and methods from a parent class.

A useful design question is:

> **Is the new type a more specific version of the existing type?**

Example:

**GraduateStudent IS A Student**

This suggests inheritance may make sense.


### Example: Inheritance and `super()`

`GraduateStudent` inherits from `Student`.

`super()` allows the child class to reuse initialization from the parent class.


````python
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

    def introduce(self):
        return f"My name is {self.name}, and my major is {self.major}."


class GraduateStudent(Student):
    def __init__(self, name, major, advisor):
        super().__init__(name, major)
        self.advisor = advisor


grad = GraduateStudent("Sarah", "Data Science", "Dr. Smith")

print(grad.introduce())
print("Advisor:", grad.advisor)
````

## 15. Inheritance vs. Composition

Two important relationships are:

### IS-A → Inheritance

Examples:

- A dog **IS AN** animal.
- A graduate student **IS A** student.
- An electric car **IS A** car.

### HAS-A → Composition

Examples:

- A laptop **HAS A** battery.
- An electric car **HAS A** battery.
- A customer **HAS** accounts.

**Inheritance → IS-A**  
**Composition → HAS-A**


## 16. Composition

**Composition** means that one object contains another object.

For example, an `ElectricCar` can contain a `Battery` object.

The car and battery have different responsibilities, so using separate classes can keep the design organized.


### Example: Composition

Here, the `ElectricCar` object contains a `Battery` object.


````python
class Battery:
    def __init__(self, size=40):
        self.size = size

    def describe(self):
        return f"{self.size}-kWh battery"


class ElectricCar:
    def __init__(self, model):
        self.model = model
        self.battery = Battery()


car = ElectricCar("Leaf")

print(car.model)
print(car.battery.describe())
````

## 17. Method Overriding — Preview

Sometimes a child class needs a different version of a method that already exists in the parent class.

This is called **method overriding**.

Example:

- `Car.fuel_type()` may return `"Gasoline"`.
- `ElectricCar.fuel_type()` may return `"Electricity"`.

For this preview, focus only on recognizing the idea.
