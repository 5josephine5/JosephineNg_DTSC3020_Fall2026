# Part A — Week 3 Review

## 1. Choosing a Data Structure

Python provides different data structures for different needs.

- **List** — ordered and mutable; useful when values may change.
- **Tuple** — ordered and immutable; useful for fixed values.
- **Set** — unordered collection of unique values; useful for removing duplicates or checking membership.
- **Dictionary** — stores information as key-value pairs.

Choosing the right data structure makes code easier to organize and understand.


## 2. Lists and Slicing

A **list** is an ordered and mutable collection.

Because lists are mutable, you can:

- update an item,
- add an item with `append()`,
- insert an item with `insert()`,
- remove an item with methods such as `pop()`.

Python indexing begins at `0`.

### Slicing

The general slicing pattern is:

`sequence[start:stop:step]`

The `start` position is included, but the `stop` position is not included.


### Example: Modify and Slice a List

Run the cell below and observe how the list changes and how slicing selects only part of it.


```python
scores = [88, 92, 79, 95, 100]

scores[2] = 85
scores.append(98)

print("Updated scores:", scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("Slice [1:4]:", scores[1:4])
```

## 3. Tuples

A **tuple** is an ordered and immutable collection.

The key difference between a list and a tuple is **mutability**:

- A list can be changed after it is created.
- A tuple cannot be changed in place after it is created.

Tuples are useful for fixed values such as coordinates, dates, or RGB colors.

A single-item tuple needs a comma, for example: `(1,)`.


### Example: List vs. Tuple

The list can be changed, but the tuple is intended to remain fixed.


```python
scores = [80, 90, 95]
coordinate = (3, 5)

scores[0] = 85
scores.append(100)

print("List:", scores)
print("Tuple:", coordinate)

# coordinate[0] = 4   # This would cause an error
```

## 4. Sets and Booleans

A **set** stores unique values.

Sets are useful for:

- removing duplicates,
- checking whether an item is present,
- comparing groups.

The `in` operator performs a membership check and returns a Boolean value: `True` or `False`.

Common set operations include:

- union: `|`
- intersection: `&`
- difference: `-`


### Example: Remove Duplicates and Check Membership

Run the cell below and notice that duplicate values are removed when the list is converted to a set.


```python
tags = ["AI", "Python", "AI", "Data", "Python"]

unique_tags = set(tags)

print("Unique tags:", unique_tags)
print("Python" in unique_tags)
print("Java" in unique_tags)
```

## 5. Dictionaries

A **dictionary** stores information as **key-value pairs**.

For example:

`{"name": "Alice", "score": 95}`

Here:

- `"name"` and `"score"` are keys.
- `"Alice"` and `95` are values.

Use square brackets to access a value by its key.

You can also add or update dictionary values after the dictionary is created.

The `.get()` method can be useful when a key might not exist.


### Example: Access and Update a Dictionary

Run the cell and observe how values are accessed and updated using keys.


```python
student = {
    "name": "Alice",
    "major": "Data Science",
    "score": 95
}

print(student["name"])

student["score"] = 98
student["email"] = "alice@unt.edu"

print(student)
print(student.get("email", "No email found"))
```

## 6. List Comprehension

A **list comprehension** is a concise way to create a new list from an existing sequence.

For example:

`[n * n for n in numbers]`

creates a new list containing the square of each value.

A condition can also be added to keep only selected values.


### Example: Transform and Filter a List

The first list comprehension transforms each number.  
The second keeps only scores that meet a condition.


```python
numbers = [1, 2, 3, 4]
squares = [n * n for n in numbers]

scores = [55, 72, 88, 40, 95]
passing_scores = [score for score in scores if score >= 70]

print("Squares:", squares)
print("Passing scores:", passing_scores)
```

## 7. Decision Making with `if`, `elif`, and `else`

An `if` statement runs a block of code only when its condition is `True`.

Use:

- `if` for the first condition,
- `elif` to check another condition,
- `else` when none of the earlier conditions are true.

Python requires a colon (`:`) after the condition, and the code block must be indented.


### Example: Grade Decision

Run the cell, then change the score and observe how the selected branch changes.


```python
score = 88

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "Below B"

print("Grade:", grade)
```

## 8. `for` Loops

A **`for` loop** repeats a block of code for each item in a sequence.

Python uses the keyword `in` to move through the sequence one item at a time.

Lists, strings, and dictionaries are examples of objects that can be processed with a `for` loop.


### Example: Loop Through a List

The loop runs once for each student in the list.


```python
students = ["Alice", "Ben", "Carla"]

for student in students:
    print("Welcome,", student)
```

## 9. The `range()` Function

`range()` generates a sequence of integers and is commonly used with `for` loops.

Three common forms are:

- `range(stop)`
- `range(start, stop)`
- `range(start, stop, step)`

The `stop` value is **not included**.


### Example: Using `range()`

Run the cell and compare the three sequences.


```python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(1, 10, 2)))
```

## 10. Switch-Case Style Logic

Some programming languages use a traditional `switch-case` statement.

In Python, similar decision logic can be written in different ways, including:

- `if-elif-else`
- `match-case` in Python 3.10 and later
- dictionary-based lookup for simple cases

For now, you only need to recognize the purpose: **choose an action based on a value**.


```python
menu_choice = "2"

match menu_choice:
    case "1":
        print("View grades")
    case "2":
        print("Submit assignment")
    case _:
        print("Unknown option")
```

## 11. Exceptions and `try-except`

An **exception** is a runtime error that occurs while a program is running.

Examples include:

- `ZeroDivisionError` — dividing by zero
- `ValueError` — using an inappropriate value, such as converting `"abc"` to an integer

A `try-except` block allows a program to handle an exception instead of stopping unexpectedly.

- Code that might fail goes inside `try`.
- The matching `except` block runs if that exception occurs.


### Example: Handle a Runtime Error

The code below attempts to divide by zero. Instead of crashing, the program handles the error.


```python
a = 10
b = 0

try:
    result = a / b
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```
