# Part A — Week 2 Review

## 1. Variables, Data Types, and Dynamic Typing

A **variable** is a name that refers to a value.

Python determines the data type from the value assigned to the variable. Common types include:

- `int` — whole numbers
- `float` — decimal numbers
- `str` — text

Python is **dynamically typed**, which means the same variable can refer to values of different types during program execution.

Use `type()` to check the data type of a value or variable.


### Example: Checking Data Types

Run the cell below and notice how the value and data type change even though the variable name stays the same.


```python
value = 10
print(value, type(value))

value = "Python"
print(value, type(value))

value = 3.14
print(value, type(value))
```

## 2. Identifiers and Type Conversion

An **identifier** is a name used for variables, functions, classes, modules, and other Python objects.

A valid Python identifier:

- can contain letters, digits, and underscores,
- cannot start with a digit,
- cannot contain spaces or symbols such as `-` or `$`,
- cannot be a Python keyword such as `if`, `for`, or `class`.

For variables and functions in this course, use meaningful `snake_case` names such as `student_score`.

### Type Conversion

Python can convert values between numeric types.

- `int()` converts a value to an integer.
- `float()` converts a value to a floating-point number.

Remember: `int(7.7)` becomes `7`. It **truncates** the decimal part rather than rounding.


### Example: Type Conversion

Run the cell and compare the original and converted values.


```python
score = 7.7

print("Original:", score)
print("As int:", int(score))
print("7 as float:", float(7))
```

## 3. Strings

A **string** is a sequence of characters used to represent text.

Important Week 2 string ideas include:

- Indexing: access one character using its position.
- Slicing: access part of a string.
- Immutability: individual characters cannot be changed in place.
- String methods: functions such as `.lower()`, `.upper()`, and `.strip()` help process text.

Python indexing starts at `0`.


### Example: String Indexing and Methods

Run the cell and observe how indexing, slicing, and string methods work.


```python
word = "Python"

print("First character:", word[0])
print("Last character:", word[-1])
print("First three characters:", word[0:3])

text = "   PYTHON Programming   "
print(text.strip())
print(text.lower())
```

## 4. Operators

Operators tell Python to perform actions on values.

### Arithmetic Operators
`+` addition  
`-` subtraction  
`*` multiplication  
`/` division  
`//` floor division  
`%` modulus (remainder)  
`**` exponentiation

### Comparison Operators
`==`, `!=`, `>`, `<`, `>=`, `<=`

A comparison produces `True` or `False`.

### Logical Operators
`and`, `or`, `not`

Python also follows **operator precedence**. Parentheses can be used to make the intended order clear.


### Example: Operators

Run the cell and compare the calculation and comparison results.


```python
a = 10
b = 3

print("Remainder:", a % b)
print("Exponent:", 2 ** 3)

score = 85
print("At least 70?", score >= 70)

print(2 + 3 * 4)
print((2 + 3) * 4)
```

## 5. Variable Scope

**Scope** describes where a variable can be seen and used.

For this review, focus on:

- **Global variable:** created outside a function.
- **Local variable:** created inside a function and normally used only inside that function.

Python also has built-in names such as `print()`, `len()`, and `sum()`.


### Example: Global and Local Variables

In this example, `course` is global and `message` is local to the function.


```python
course = "DTSC 3020"

def show_message():
    message = "Welcome!"
    print(message)
    print(course)

show_message()
print(course)
```

## 6. Random Numbers

Python's `random` module can generate **pseudorandom** values.

Some basic functions from Week 2 include:

- `random.random()` — returns a floating-point value between 0 and 1.
- `random.randint(a, b)` — returns a random integer from `a` through `b`, including both endpoints.
- `random.choice(sequence)` — selects one random item.

`random.seed()` is useful when you want the same pseudorandom sequence to be reproducible.


### Example: Reproducible Random Values

Run the cell more than once. Because the seed is reset to the same value, the same sequence is produced.


```python
import random

random.seed(7)
print(random.randint(1, 10))
print(random.randint(1, 10))

random.seed(7)
print(random.randint(1, 10))
print(random.randint(1, 10))
```

## 7. Lists

A **list** is an **ordered and mutable** collection.

- **Ordered** means items have positions (indexes).
- **Mutable** means the list can be changed after it is created.
- Lists use square brackets: `[ ]`

You can access list items by index, update items, and add new items with methods such as `append()`.


### Example: A List

Run the cell below. Then change one score or add another score.


```python
scores = [88, 92, 79, 95]

print("First score:", scores[0])
print("Last score:", scores[-1])

scores[2] = 85
scores.append(100)

print("Updated scores:", scores)
```

## 8. Slicing

**Slicing** selects part of a sequence.

The basic pattern is:

`sequence[start:stop]`

The `start` index is included, but the `stop` index is **not included**.

For example, `items[1:4]` selects the items at indexes `1`, `2`, and `3`.


```python
students = ["Alice", "Ben", "Carla", "David", "Eva"]

print(students[1:4])
print(students[:3])
```

## 9. Tuples

A **tuple** is an **ordered and immutable** collection.

Tuples use parentheses: `( )`

The main beginner-level difference is:

- **List:** mutable — its items can be changed.
- **Tuple:** immutable — its items cannot be changed in place.

Tuples are useful for values that should remain fixed, such as coordinates.


```python
scores = [80, 90, 95]
coordinate = (3, 5)

scores[0] = 85       # allowed because a list is mutable
print(scores)

print(coordinate)
# coordinate[0] = 4  # would cause an error
```

## 10. Sets

A **set** is an unordered collection of **unique values**.

Sets are especially useful when you want to:

- remove duplicates,
- test whether an item is present,
- compare groups of values.

Sets use curly braces: `{ }`


```python
tags = ["AI", "Python", "AI", "Data", "Python"]

unique_tags = set(tags)

print("Original:", tags)
print("Unique:", unique_tags)
print("Python" in unique_tags)
```

## 11. Dictionaries

A **dictionary** stores information as **key-value pairs**.

For example, a student record might use keys such as `"name"` and `"score"`.

Dictionaries use curly braces with a colon between each key and value:

`{"name": "Alice", "score": 95}`

A value can be accessed using its key.


```python
student = {
    "name": "Alice",
    "major": "Data Science",
    "score": 95
}

print(student["name"])
print(student["score"])
```

## 12. List Comprehension — Preview

A **list comprehension** is a concise way to create a new list from an existing sequence.

For example:

`[n * n for n in numbers]`

creates a new list containing the square of each value in `numbers`.

You will practice list comprehensions in more detail during Week 3.


```python
numbers = [1, 2, 3, 4]

squares = [n * n for n in numbers]

print(squares)
```
