# 1. Working with Files, Folders, and Paths

The `os` module helps Python interact with the operating system.

Important tools from this lesson:

| Tool | Purpose |
|---|---|
| `os.getcwd()` | Show the current working directory |
| `os.listdir()` | List files and folders |
| `os.makedirs()` | Create a folder |
| `os.path.exists()` | Check whether a path exists |
| `os.path.join()` | Build a portable path |

### Why use `os.path.join()`?
Different operating systems may represent paths differently. `os.path.join()` helps build paths that are easier to reuse across environments.


````python
import os

print("Current directory:", os.getcwd())
print("Some items in this directory:", os.listdir(".")[:5])

folder = "quiz_prep_demo"
os.makedirs(folder, exist_ok=True)

file_path = os.path.join(folder, "notes.txt")
print("Portable path:", file_path)
print("Does the file exist yet?", os.path.exists(file_path))
````

## Key idea

Before reading, copying, renaming, or overwriting a file, it is good practice to check that the path is correct and whether the file exists.


# 2. Reading and Writing Text Files

Python commonly uses:

```python
with open(path, mode, encoding="utf-8") as file:
    ...
```

Using `with open(...)` is recommended because Python automatically closes the file when the block finishes.

## Important file modes

| Mode | Meaning | Important behavior |
|---|---|---|
| `"r"` | Read | File must already exist |
| `"w"` | Write | Overwrites existing content |
| `"a"` | Append | Adds new content to the end |
| `"x"` | Create | Fails if the file already exists |
| `"rb"` / `"wb"` | Binary read/write | Used for non-text files |

### Important
Opening an existing file with `"w"` **overwrites** its old content. Use `"a"` when you want to add to the end.


````python
# Write a file
with open(file_path, "w", encoding="utf-8") as file:
    file.write("Python file practice\n")
    file.write("Files can store data outside the program.\n")

# Read the whole file
with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

````python
# Append one more line
with open(file_path, "a", encoding="utf-8") as file:
    file.write("This line was appended later.\n")

with open(file_path, "r", encoding="utf-8") as file:
    print(file.read())
````

# 3. File Reading Methods

Different methods are useful for different situations.

| Method | Result / behavior |
|---|---|
| `read()` | Reads the entire file as one string |
| `readline()` | Reads one line at a time |
| `readlines()` | Reads all lines into a list |
| `for line in file` | Processes the file line by line |

For large files, line-by-line processing is usually more memory-friendly than reading the whole file at once.


````python
with open(file_path, "r", encoding="utf-8") as file:
    for line in file:
        print("Line:", line.strip())
````

# 4. Copying, Renaming, and ZIP Files

Python can automate basic file-management tasks.

Important tools from this lesson:

- `shutil.copy(source, destination)` — copy a file
- `os.rename(old_path, new_path)` — rename a file
- `zipfile.ZipFile(...)` — create or inspect ZIP archives
- `namelist()` — show files stored inside a ZIP archive

Before changing files, always make sure the source and destination paths are correct.


````python
import shutil
import os

copy_path = os.path.join(folder, "notes_copy.txt")
renamed_path = os.path.join(folder, "notes_backup.txt")

shutil.copy(file_path, copy_path)

if os.path.exists(copy_path):
    os.rename(copy_path, renamed_path)

print("Backup exists:", os.path.exists(renamed_path))
````

````python
import zipfile

zip_path = os.path.join(folder, "practice_archive.zip")

with zipfile.ZipFile(zip_path, "w") as zip_file:
    zip_file.write(file_path, arcname="notes.txt")

with zipfile.ZipFile(zip_path, "r") as zip_file:
    print("Files in archive:", zip_file.namelist())
````

# 5. Exceptions

An **exception** is a problem detected while a program is running.

An unhandled exception can stop the current execution.

A **syntax error** is different: it means Python cannot correctly parse the code, such as when required punctuation is missing.

## Common exceptions from this lesson

| Exception | Example cause |
|---|---|
| `FileNotFoundError` | Trying to read a file that does not exist |
| `ValueError` | Trying to convert `"eighty"` to a number |
| `ZeroDivisionError` | Dividing by zero |
| `TypeError` | Using incompatible data types together |
| `PermissionError` | Trying to access a file without required permission |

### Reading an error message
A useful first step is to read the **last line** of the error message. It usually gives the exception type and a description of the problem.


# 6. `try`, `except`, `else`, and `finally`

| Block | When it runs |
|---|---|
| `try` | Attempts code that may fail |
| `except` | Handles a matching exception raised in `try` |
| `else` | Runs if the `try` block finishes without an exception |
| `finally` | Runs whether or not an exception occurs |

Catch the **specific exception** you expect whenever possible.


````python
missing_path = os.path.join(folder, "missing_file.txt")

try:
    with open(missing_path, "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("The file was not found.")
````

````python
for raw_score in ["88", "eighty-eight"]:
    try:
        numeric_score = float(raw_score)
    except ValueError:
        print("Cannot convert:", raw_score)
    else:
        print("Converted score:", numeric_score)
    finally:
        print("Finished checking:", raw_score)
````

## Important distinction

- `except` runs only when the matching exception occurs.
- `else` runs only when the `try` block succeeds.
- `finally` runs in both the success and failure cases.


# 7. Raising an Exception

A program can intentionally raise an exception when the input is unsuitable.

Example:

```python
raise ValueError("Provide at least one score.")
```

This is useful when a function has clear requirements and should reject invalid input.


````python
def calculate_average(scores):
    if not scores:
        raise ValueError("Provide at least one score.")

    if any(score < 0 or score > 100 for score in scores):
        raise ValueError("Scores must be between 0 and 100.")

    return sum(scores) / len(scores)

print(calculate_average([80, 90, 100]))
````

# 8. Testing Code

Code can run without producing an exception and still produce the **wrong answer**.

Testing compares an **actual result** with an **expected result**.

Important test categories:

| Category | Example for `calculate_average()` |
|---|---|
| Normal | `[80, 90, 100]` |
| Single value | `[75]` |
| Boundary | `[0, 100]` |
| Invalid | `[]` or `[105]` |

Good testing does not use only one normal example. Different cases help reveal different mistakes.


# 9. Using `assert`

An assertion checks whether a condition is true.

```python
assert condition
```

If the condition is false, Python raises an `AssertionError`.

Assertions are useful for quick tests. They should not be the only method used to validate user input.


````python
assert calculate_average([80, 90, 100]) == 90
assert calculate_average([75]) == 75
assert calculate_average([0, 100]) == 50

print("All quick checks passed.")
````

# 10. Basic `unittest`

Python's built-in `unittest` module can organize several tests.

Two important methods from this lesson:

- `self.assertEqual(actual, expected)` — checks whether two values are equal.
- `self.assertRaises(ExceptionType)` — checks whether the expected exception occurs.

You should understand the purpose of these methods and be able to recognize what they are testing.


````python
import unittest

class TestCalculateAverage(unittest.TestCase):
    def test_normal_scores(self):
        self.assertEqual(calculate_average([80, 90, 100]), 90)

    def test_empty_scores(self):
        with self.assertRaises(ValueError):
            calculate_average([])

suite = unittest.defaultTestLoader.loadTestsFromTestCase(TestCalculateAverage)
unittest.TextTestRunner(verbosity=2).run(suite)
````
