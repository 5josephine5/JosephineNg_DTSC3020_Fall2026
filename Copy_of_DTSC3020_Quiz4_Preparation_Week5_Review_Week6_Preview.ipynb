# Part A — Week 5 Review
## 1. `if`, `elif`, and `else`

Conditional statements allow a program to make decisions.

- `if` checks the first condition.
- `elif` checks another condition if the earlier condition was false.
- `else` runs when none of the earlier conditions are true.

Python uses indentation to show which statements belong to each branch.


### Example: Making a Decision

Change the value of `score` and run the cell again to see which branch executes.


```python
score = 86

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
else:
    print("Below B")
```

## 2. Nested `if` Statements

A **nested `if` statement** is an `if` statement placed inside another `if` block.

This is useful when a second condition should be checked only after a first condition is true.


### Example: Nested Decision

The second condition is checked only when `age >= 18`.


```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Too young")
```

## 3. `for` Loops

A `for` loop repeats a block of code once for each item in a sequence.

The loop variable takes one value from the sequence during each iteration.


### Example: Loop Through a List

The loop below runs once for each student.


```python
students = ["Alice", "Ben", "Carla"]

for student in students:
    print("Hello,", student)
```

## 4. `range()`

`range()` generates a sequence of integers and is commonly used with `for` loops.

Common forms:

- `range(stop)`
- `range(start, stop)`
- `range(start, stop, step)`

The **stop value is not included**.


### Example: Using `range()`

Compare the three sequences below.


```python
print(list(range(5)))
print(list(range(2, 6)))
print(list(range(1, 10, 2)))
```

## 5. Switch-Case Style Logic

Python can use `match-case` for switch-case style logic.

Its purpose is to choose an action based on the value being matched.

The wildcard case `_` can be used as a default option when no earlier case matches.


### Example: `match-case`

Run the cell, then change `menu_choice`.


```python
menu_choice = "2"

match menu_choice:
    case "1":
        print("View grades")
    case "2":
        print("Submit assignment")
    case _:
        print("Unknown option")
```

## 6. Exceptions and Runtime Errors

An **exception** is an error that occurs while a program is running.

Examples include:

- `ZeroDivisionError` — dividing by zero
- `ValueError` — using an inappropriate value, such as `int("abc")`
- `IndexError` — requesting a list index that does not exist
- `KeyError` — requesting a dictionary key that does not exist


## 7. `try-except`

A `try-except` block allows a program to handle an exception instead of stopping unexpectedly.

- Put code that might fail inside `try`.
- Use `except` to define what should happen when a particular exception occurs.


### Example: Handling Division by Zero

Instead of stopping with an error, the program prints a message.


```python
a = 10
b = 0

try:
    result = a / b
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

## 8. What Is a Function?

A **function** is a reusable block of code designed to perform a specific task.

Functions help:

- organize code,
- avoid repeating the same code,
- make programs easier to read and maintain.

In Python, a function is commonly defined with `def`.


### Example: Define and Call a Function

Defining a function creates it. Calling the function runs its code.


```python
def greet():
    print("Hello, DTSC 3020!")

greet()
```

## 9. Parameters, Arguments, and Return Values

A **parameter** is a variable listed in a function definition.

An **argument** is the actual value passed to the function when it is called.

A function can use `return` to send a result back to the part of the program that called it.


### Example: Parameter, Argument, and Return

Here, `number` is a parameter and `5` is the argument.


```python
def square(number):
    return number * number

result = square(5)
print(result)
```

## 10. Positional, Keyword, and Default Arguments

Python functions can receive arguments in different ways.

- **Positional argument:** matched by position.
- **Keyword argument:** matched by parameter name.
- **Default argument:** uses a predefined value if no value is supplied.


### Example: Default and Keyword Arguments

The first call uses the default greeting.  
The second call supplies the greeting by keyword.


```python
def welcome(name, greeting="Hello"):
    print(greeting, name)

welcome("Alice")
welcome(name="Ben", greeting="Hi")
```

For now, focus only on their basic purpose:

## 12. `zip()` and `repr()`

Two useful built-in functions are:

- `zip()` — combines items from multiple sequences position by position.
- `repr()` — returns a representation of a value that is useful for inspecting or debugging it.


### Example: `zip()`

The first name is paired with the first score, the second with the second, and so on.


```python
names = ["Alice", "Ben", "Carla"]
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(name, score)
```

## 13. `eval()` and `exec()`

You only need the big idea for now:

- `eval()` evaluates a Python expression provided as text.
- `exec()` executes Python code provided as text.

Because they can run text as Python code, they can be **unsafe with untrusted input**.

Do not use them with text from unknown or untrusted sources.


## 14. Regular Expressions

A **regular expression (regex)** is a pattern used to search or work with text.

Regex can be used for tasks such as:

- searching for a pattern,
- matching text,
- extracting information,
- replacing text.

Python commonly uses the `re` module for regular expressions.


## 15. Recursion

**Recursion** happens when a function calls itself.

A recursive function needs a **base case** that tells the function when to stop.

Without a proper stopping condition, the function can continue calling itself until an error occurs.


### Example: Simple Recursion

The base case stops the function when `n` reaches `0`.


```python
def countdown(n):
    if n == 0:
        print("Done!")
        return

    print(n)
    countdown(n - 1)

countdown(3)
```

## 16. Lambda Expressions

A **lambda expression** creates a small anonymous function.

A lambda is often useful when a short function is needed temporarily.

General form:

`lambda parameters: expression`


### Example: Lambda

This lambda takes one value and returns its square.


```python
square = lambda x: x * x

print(square(4))
```
