# Part A — Week 6 Review
## 1. Defining and Calling Functions

A **function** is a reusable block of code that performs a specific task.

Functions help you:

- avoid repeating code,
- organize a program into smaller pieces,
- make code easier to read and maintain.

In Python, a function is commonly defined with `def`.

A function does not run when it is defined. It runs when it is **called**.


### Example: Define and Call a Function

Change the name passed to `greet()` and run the cell again.


````python
def greet(name):
    print("Hello,", name)

greet("Alice")
````

## 2. Parameters, Arguments, and Return Values

A **parameter** is a variable listed in the function definition.

An **argument** is the actual value passed to the function when it is called.

A function can use `return` to send a result back to the part of the program that called it.

Remember:

- `print()` displays a value.
- `return` sends a value back from the function.


### Example: Parameter, Argument, and Return

Here, `number` is a parameter and `5` is an argument.


````python
def square(number):
    return number * number

result = square(5)
print(result)
````

## 3. Positional, Keyword, and Default Arguments

Python functions can receive arguments in different ways.

- **Positional argument:** matched by position.
- **Keyword argument:** matched by parameter name.
- **Default argument:** uses a predefined value when no value is supplied.

The order matters for positional arguments, while keyword arguments explicitly name the parameter.


### Example: Argument Types

Compare the three calls below.


````python
def welcome(name, greeting="Hello"):
    print(greeting, name)

welcome("Alice")                         # positional + default
welcome("Ben", "Hi")                    # positional
welcome(name="Carla", greeting="Hey")   # keyword
````

## 4. `*args` and `**kwargs`

Sometimes you do not know in advance how many arguments a function will receive.

- `*args` collects extra **positional arguments**.
- `**kwargs` collects extra **keyword arguments**.

Inside the function:

- `args` behaves like a tuple.
- `kwargs` behaves like a dictionary.


### Example: `*args`

This function can receive any number of scores.


````python
def average(*scores):
    return sum(scores) / len(scores)

print(average(80, 90, 100))
````

### Example: `**kwargs`

This function receives information using keyword arguments.


````python
def show_student(**info):
    for key, value in info.items():
        print(key, ":", value)

show_student(name="Alice", major="Data Science")
````

## 5. `zip()` and `repr()`

Two useful built-in functions are:

- `zip()` — combines items from multiple sequences position by position.
- `repr()` — returns a representation of a value that is useful for inspecting or debugging it.

`zip()` is especially useful when related lists should be processed together.


### Example: `zip()`

The first name is paired with the first score, the second with the second, and so on.


````python
names = ["Alice", "Ben", "Carla"]
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(name, score)
````

### Example: `repr()`

Notice that `repr()` makes special characters such as `\n` visible in the representation.


````python
text = "Hello\nWorld"

print(text)
print(repr(text))
````

## 6. `eval()` and `exec()`

- `eval()` evaluates a Python **expression** stored as text.
- `exec()` executes Python **statements or code** stored as text.

These functions can be dangerous when the text comes from an unknown or untrusted source because the text may contain code you did not intend to run.

For this course, focus on understanding their purpose and their potential risk.


## 7. Regular Expressions

A **regular expression (regex)** is a pattern used to search or work with text.

Python commonly uses the `re` module for regular expressions.

Common operations include:

- `re.search()` — finds a pattern anywhere in the text,
- `re.match()` — checks for a match at the beginning,
- `re.findall()` — extracts all matching pieces,
- `re.sub()` — replaces matching text.


### Example: Find Numbers in Text

The pattern `\d+` means one or more digits.


````python
import re

text = "Quiz scores: 85, 92, and 78"
numbers = re.findall(r"\d+", text)

print(numbers)
````

### Example: Replace Text with `re.sub()`

This replaces every sequence of digits with the word `"score"`.


````python
import re

text = "Student 1 earned 95 points."
updated = re.sub(r"\d+", "score", text)

print(updated)
````

## 8. Recursion

**Recursion** happens when a function calls itself.

A recursive function needs a **base case** that tells the function when to stop.

Without a correct base case, the function may continue calling itself until Python raises an error.


### Example: Simple Recursion

The base case stops the function when `n` reaches `0`.


````python
def countdown(n):
    if n == 0:
        print("Done!")
        return

    print(n)
    countdown(n - 1)

countdown(3)
````

## 9. Lambda Expressions

A **lambda expression** creates a small anonymous function.

General form:

`lambda parameters: expression`

A lambda is useful when a short function is needed temporarily.


### Example: Lambda

This lambda takes one value and returns its square.


````python
square = lambda x: x * x

print(square(4))
````

## 10. Putting Functions Together

Functions are most useful when they organize a practical task into reusable steps.

In the example below:

1. `calculate_average()` performs a calculation.
2. `get_status()` uses the result to make a decision.


````python
def calculate_average(scores):
    return sum(scores) / len(scores)

def get_status(average):
    if average >= 70:
        return "Passing"
    return "Needs Improvement"

student_scores = [75, 82, 68]
avg = calculate_average(student_scores)

print("Average:", avg)
print("Status:", get_status(avg))
````

## 11. What Is a Module?

A **module** is a Python file that contains reusable code.

A module can contain:

- variables,
- functions,
- classes,
- and other Python statements.

Instead of writing every piece of code from scratch, you can import a module and use code that already exists.

Python includes many modules in its **standard library**, and you can also create your own modules.


### Example: Using a Standard-Library Module

The `math` module contains mathematical functions and constants.

After importing it, we can access its contents using dot notation.


````python
import math

print(math.sqrt(25))
print(math.pi)
````

## 12. Common Import Styles

Python supports several common ways to import code.

### 1. Import the entire module

```python
import math
print(math.sqrt(16))
```

You use the module name before the function.

### 2. Import with an alias

```python
import math as m
print(m.sqrt(16))
```

An **alias** gives the module a shorter or more convenient name.

### 3. Import a specific item

```python
from math import sqrt
print(sqrt(16))
```

Here, you can use `sqrt()` directly without writing `math.sqrt()`.


### Try the Three Import Styles

Run this cell and compare the syntax.


````python
import math
print(math.ceil(4.2))

import math as m
print(m.floor(4.8))

from math import sqrt
print(sqrt(81))
````

## 13. Why Use Modules?

Modules help make programs more organized.

Imagine a program with hundreds of lines of code. Instead of putting everything in one file, you can separate related tasks.

For example:

- `calculations.py` could contain math functions,
- `messages.py` could contain formatting functions,
- `main.py` could import and use both.

This makes code easier to:

- read,
- test,
- update,
- and reuse.


## 14. The `datetime` Module

Python's `datetime` module provides tools for working with dates and times.

Common objects include:

- `date` — represents a calendar date,
- `time` — represents a time of day,
- `datetime` — represents both date and time,
- `timedelta` — represents a duration or difference between dates/times.

You do not need to memorize every method, but you should recognize how these objects are created and used.


### Example: Working with Dates


````python
from datetime import date

today = date.today()
print("Today:", today)

semester_start = date(2026, 8, 24)
print("Semester start:", semester_start)
````

## 15. Date Calculations

Date objects can be compared and subtracted.

Subtracting one date from another returns a `timedelta` object.

The `.days` attribute gives the number of days in that difference.


### Example: Days Between Two Dates


````python
from datetime import date

start = date(2026, 9, 1)
end = date(2026, 9, 22)

difference = end - start

print(difference)
print("Number of days:", difference.days)
````

## 16. Adding or Subtracting Time with `timedelta`

A `timedelta` can also be used to move forward or backward in time.

For example, you can calculate a date one week from today.


````python
from datetime import date, timedelta

today = date.today()
one_week = timedelta(days=7)

next_week = today + one_week

print("Today:", today)
print("One week later:", next_week)
````

## 17. Modules vs. Packages vs. Libraries

These terms are related, but they are not exactly the same.

### Module
Usually a single Python file containing reusable code.

Example:

`calculator.py`

### Package
A collection of related Python modules organized in a directory.

For example, a package might contain:

- `statistics.py`
- `visualization.py`
- `cleaning.py`

### Library
A broader collection of reusable tools. A library may include many modules and packages.

Examples of well-known Python libraries include NumPy, pandas, and Matplotlib.

A simple way to remember the relationship is:

**module → one file**  
**package → collection of related modules**  
**library → larger collection of reusable functionality**


## 18. Creating Your Own Module

You can create a module simply by placing reusable Python code in a `.py` file.

Suppose a file named `my_utils.py` contains:

```python
def double(number):
    return number * 2

def greet(name):
    return f"Hello, {name}!"
```

Another Python file in the same project can import those functions:

```python
from my_utils import double, greet

print(double(5))
print(greet("Alice"))
```

The important idea is that the functions are written once and then reused elsewhere.


## 19. Why Separate Code into Multiple Files?

As programs become larger, keeping everything in one file can become difficult.

Modular programming helps separate responsibilities.

For example:

```text
project/
│
├── main.py
├── calculations.py
└── messages.py
```

`main.py` can import functions from the other files.

This structure makes it easier to understand which file is responsible for each part of the program.


## 20. What Is a Virtual Environment?

A **virtual environment** is an isolated Python environment created for a specific project.

Different projects may need different versions of the same package.

For example:

- Project A may require one version of a library.
- Project B may require a newer version.

A virtual environment helps prevent these project dependencies from interfering with each other.


## 21. Basic Virtual Environment Workflow

You do not need to memorize every operating-system-specific detail yet, but you should recognize the typical workflow.

### Create a virtual environment

```bash
python -m venv myenv
```

### Activate it on Windows

```bash
myenv\Scripts\activate
```

### Activate it on macOS/Linux

```bash
source myenv/bin/activate
```

### Install a package

```bash
pip install package_name
```

Once the environment is activated, packages installed with `pip` are associated with that environment.


## 22. Installing and Importing Are Different

These two steps are related but different.

### Installing

```bash
pip install pandas
```

This makes the package available in your environment.

### Importing

```python
import pandas
```

This makes the installed package available to your Python program.

So:

**install → makes the package available to the environment**  
**import → makes the package available inside your code**


## 23. Libraries in Real Programs

Libraries give programmers access to reusable functionality without building everything from scratch.

For example:

- `datetime` helps with dates and times,
- `math` provides mathematical operations,
- pandas helps work with tabular data,
- Matplotlib helps create visualizations.

A programmer can combine:

- Python's built-in features,
- standard-library modules,
- external libraries,
- and their own functions/modules

to build a larger program.


## 24. Putting the Ideas Together

Consider this simple example:

```python
from datetime import date

def build_message(name):
    today = date.today()
    return f"Hello {name}. Today is {today}."

print(build_message("Alice"))
```

This program combines two ideas:

1. It imports reusable code from the `datetime` module.
2. It defines its own reusable function.

This is the basic idea behind **modular programming**: combining small, reusable pieces of code to build a larger program.
