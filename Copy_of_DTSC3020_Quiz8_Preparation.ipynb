## Lesson 8: Files, Exceptions, and Testing

### 1. File Modes and File Reading

| Item | Meaning |
|---|---|
| `"r"` | Read an existing file |
| `"w"` | Write and overwrite existing content |
| `"a"` | Add new content to the end |
| `"x"` | Create a new file; fails if it already exists |
| `with open(...)` | Opens a file and closes it automatically |
| `read()` | Reads the whole file |
| `readline()` | Reads one line |
| `readlines()` | Reads all lines into a list |
| `for line in file` | Processes the file line by line |

**Important:** `csv.reader(...)` reads values as **strings**.


````python
# Example
with open("notes.txt", "a", encoding="utf-8") as file:
    file.write("New line\n")
````

### 2. JSON Basics

Valid JSON uses:
- double quotes around keys and string values;
- no Python comments;
- standard JSON syntax.

Example of valid JSON:

```json
{"name": "Alice", "score": 90}
```


### 3. Exceptions

Common examples:

- `FileNotFoundError` — missing file
- `ValueError` — invalid value conversion
- `ZeroDivisionError` — division by zero

In a traceback, the **last line** usually tells you the exception type and message.

`try / except / else / finally`:

- `except` → runs when the matching exception occurs
- `else` → runs when `try` succeeds
- `finally` → runs whether or not an exception occurs


````python
try:
    value = int("abc")
except ValueError:
    print("Invalid value")
````

### 4. `raise` vs `assert`

- Use **`raise`** to report invalid input.
- Use **`assert`** mainly to check programmer assumptions during development.

```python
if score < 0:
    raise ValueError("Score cannot be negative")

assert total >= 0
```

If an assertion is false, Python raises `AssertionError`.


### 5. Testing and Debugging

Good tests come from the **specification**, not from the current program output.

Important test cases:
- normal / typical
- boundary
- empty or minimal
- invalid

Example: if an A starts at **90**, then **90** is an important boundary value.

Other key ideas:
- **Black-box testing:** design tests from the specification without using the implementation.
- **Logic error:** the program runs but gives a wrong result.
- **Regression test:** added after fixing a bug to catch the same bug if it returns.
- `pytest` usually discovers test functions whose names begin with `test_`.

## Lesson 9: NumPy, Pandas, and Scientific Data Processing

### 6. NumPy Basics

A **NumPy array** is useful for efficient numerical calculations.

Important properties:

| Property | Meaning |
|---|---|
| `.shape` | Dimensions of the array |
| `.ndim` | Number of dimensions |
| `.size` | Total number of values |
| `.dtype` | Data type |

Common calculations:
- `.mean()`
- `.min()`
- `.max()`


````python
import numpy as np

scores = np.array([95, 88, 91, 76])

print(scores.size)
print(scores.mean())
````

### 7. Vectorization and Boolean Filtering

**Vectorization** means applying an operation to the whole array instead of writing a Python loop for each value.

Boolean filtering keeps values that satisfy a condition.


````python
scores = np.array([95, 88, 91, 76])

high_scores = scores[scores >= 90]
print(high_scores)
````

### 8. Two-Dimensional Arrays and `axis`

Suppose:
- each **row** = one student
- each **column** = one assignment

Then:

- `mean(axis=0)` → one mean for each **column / assignment**
- `mean(axis=1)` → one mean for each **row / student**


````python
grades = np.array([
    [90, 80, 100],
    [70, 85, 90]
])

print(grades.mean(axis=0))  # assignment means
print(grades.mean(axis=1))  # student means
````

### 9. Pandas `DataFrame`

A Pandas `DataFrame` is a table with labeled rows and columns.

Useful operations:

| Code | Purpose |
|---|---|
| `df.head()` | Preview the first rows |
| `df.shape` | Number of rows and columns |
| `df.dtypes` | Column data types |
| `df.isna().sum()` | Count missing values |
| `fillna(...)` | Replace missing values |
| `df[df["score"] >= 90]` | Filter rows |
| `groupby(...)` | Group rows and summarize them |
| `sort_values(...)` | Sort rows |


````python
import pandas as pd

df = pd.DataFrame({
    "student": ["Alice", "Ben", "Carla"],
    "score": [95, 88, None]
})

print(df.isna().sum())
````

### 10. Merge, SciPy, and Safe Data Processing

- `merge(...)` combines tables using a shared key.
- `to_csv(..., index=False)` saves a DataFrame without adding the row index as a new column.
- **SciPy** builds on NumPy and provides higher-level scientific tools such as statistics, optimization, and distance calculations.
