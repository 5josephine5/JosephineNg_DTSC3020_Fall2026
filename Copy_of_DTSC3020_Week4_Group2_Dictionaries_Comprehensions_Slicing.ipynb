# Part A — Dictionaries

A **dictionary** stores information as **key–value pairs**. Each key is a label that helps us find its value.

A dictionary uses curly brackets: `{}`. A colon separates each key from its value.

````python
# Curly brackets create a dictionary.
# Each entry has a key on the left and its value on the right.
# For example, the key "name" is connected to the value "Anna".
student = {
    "name": "Anna",
    "major": "Data Science",
    "credits": 45,
    "active": True
}

# The dictionary keeps different types of related data in one record.
# Here, the values include strings, an integer, and a Boolean.
print("Complete student record:", student)
````

### Dictionary Rules

- Keys must be unique. If a key is repeated, the later value replaces the earlier one.
- Values do not need to be unique.
- Values can be strings, numbers, lists, Booleans, or even other dictionaries.
- We access a value by its key, not by a numeric position.

### Accessing Values

Square brackets return the value for a known key. The `get()` method is safer when a key may not exist.

````python
# Place an exact key inside square brackets to access its value.
# The key "name" returns "Anna".
print("Name:", student["name"])

# The key "credits" returns the integer 45.
print("Credits:", student["credits"])

# The key "email" does not exist in this dictionary.
# get() returns our default message instead of causing a KeyError.
email = student.get("email", "Email not provided")
print("Email:", email)
````

> It would cause a `KeyError` because the key does not exist.

### Adding, Updating, and Removing Items

````python
# Step 1: Start with a fresh dictionary so this cell can be run more than once.
student = {"name": "Anna", "major": "Data Science", "credits": 45}

# Step 2: Assigning to a new key adds a new key–value pair.
student["email"] = "anna@university.edu"
print("After adding email:", student)

# Step 3: Assigning to an existing key replaces its old value.
student["credits"] = 48
print("After updating credits:", student)

# Step 4: pop() removes the key "major" and returns its value.
# We save that returned value in removed_major.
removed_major = student.pop("major")
print("After removing major:", student)
print("Removed major:", removed_major)
````

### Useful Dictionary Methods

- `keys()` returns the keys.
- `values()` returns the values.
- `items()` returns each key and value together as a pair.

````python
# In this dictionary, student names are keys and grades are values.
student_grades = {"Anna": 90, "John": 85, "Sara": 95}

# keys() shows the labels we can use to access the dictionary.
print("Keys:", student_grades.keys())

# values() shows the information connected to those keys.
print("Values:", student_grades.values())

# items() keeps each key beside its value as a pair.
print("Items:", student_grades.items())
````

### Nested Dictionaries

A dictionary can contain other dictionaries. This is useful for storing detailed records.

````python
# The outer dictionary uses product names as keys.
# Each product points to an inner dictionary with two details.
inventory = {
    "laptop": {"price": 850, "in_stock": True},
    "tablet": {"price": 300, "in_stock": False},
    "headphones": {"price": 75, "in_stock": True}
}

# One key returns the complete inner dictionary for the laptop.
print("Laptop record:", inventory["laptop"])

# Two keys are needed to reach one value in a nested dictionary.
# First choose "laptop", and then choose "price".
print("Laptop price:", inventory["laptop"]["price"])

# First choose "tablet", and then choose "in_stock".
print("Is the tablet in stock?", inventory["tablet"]["in_stock"])
````

# Part B — List Comprehensions

A **list comprehension** creates a new list in one concise line. It is useful for a simple transformation or selection.

Basic pattern:

```python
new_list = [expression for item in collection]
```

Read it as: “Create the expression for each item in the collection.”

````python
# This is the original list that we want to transform.
numbers = [1, 2, 3, 4, 5]

# Read the comprehension from left to right:
# 1. number ** 2 is the value added to the new list.
# 2. for number in numbers visits each item in the original list.
# The original list is not changed.
squares = [number ** 2 for number in numbers]

print("Original numbers:", numbers)
print("Squares:", squares)
````

### Adding a Condition

An `if` condition keeps only the items that meet a requirement.

```python
new_list = [expression for item in collection if condition]
```

````python
# This is the original list of all grades.
grades = [88, 72, 95, 60, 100, 45]

# Read this comprehension in three parts:
# 1. grade is the value placed in the new list.
# 2. for grade in grades examines each original grade.
# 3. if grade >= 70 keeps only grades that meet the condition.
passing_grades = [grade for grade in grades if grade >= 70]

print("All grades:", grades)
print("Passing grades:", passing_grades)
````

> **Pause and Predict:** What will the following comprehension produce?
>
> `short_names = [name for name in ["Ava", "Daniel", "Mia", "Robert"] if len(name) <= 3]`

````python
# Start with four names of different lengths.
names = ["Ava", "Daniel", "Mia", "Robert"]

# len(name) counts the characters in the current name.
# The condition keeps the name only when it has three or fewer characters.
short_names = [name for name in names if len(name) <= 3]

print("Original names:", names)
print("Names with 3 or fewer characters:", short_names)
````

> **Scope note:** Python also supports set and dictionary comprehensions. This introductory demo focuses on **list comprehensions**.

# Part C — Slicing

**Slicing** selects part of an ordered collection, such as a list or string.

```python
collection[start:stop:step]
```

- `start` is included.
- `stop` is not included.
- `step` controls how far Python moves each time.
- Any of the three values may be omitted.

````python
# The seven values represent daily sales from Monday through Sunday.
daily_sales = [120, 150, 90, 200, 175, 80, 130]

# [0:3] starts at index 0 and stops before index 3.
print("Indexes 0 through 2:", daily_sales[0:3])

# A missing start means "begin at the first item."
print("First four values:", daily_sales[:4])

# A missing stop means "continue through the last item."
print("Index 3 to the end:", daily_sales[3:])

# Index -2 means the second item from the end.
print("Last two values:", daily_sales[-2:])
````

### Using a Step

````python
# This ordered list makes it easy to see how the step changes a slice.
numbers = [0, 1, 2, 3, 4, 5, 6, 7]

# Both start and stop are omitted; step 2 selects every second item.
print("Every second item from index 0:", numbers[::2])

# Start at index 1, and then continue with the same step of 2.
print("Every second item from index 1:", numbers[1::2])

# A negative step moves backward and creates a reversed copy.
print("Reversed list:", numbers[::-1])
````

### Slicing a String

Strings are ordered, so the same slicing rules apply.

````python
# A string is an ordered sequence of characters, so it can be sliced.
course_code = "DTSC3020"

# [:4] starts at the beginning and stops before index 4.
print("Department code:", course_code[:4])

# [4:] starts at index 4 and continues to the end.
print("Course number:", course_code[4:])

# [-4:] is another way to select the final four characters.
print("Last four characters:", course_code[-4:])
````

> **Important:** Slicing creates a new result. It does not change the original list or string. Also remember that the `stop` position is excluded.

# Final Mixed Example — Student Results

This example combines a dictionary, a list comprehension, and slicing.

````python
# Step 1: Create a dictionary that connects each name to one grade.
student_grades = {
    "Anna": 92,
    "John": 76,
    "Sara": 88,
    "Lina": 65,
    "Reza": 95
}

# Step 2: items() provides each name and grade together.
# During each visit, name stores the key and grade stores its value.
# The if condition keeps only names connected to a grade of 80 or higher.
high_achievers = [name for name, grade in student_grades.items() if grade >= 80]

# Step 3: [:2] starts at the beginning and stops before index 2.
# Therefore, it selects the first two names from high_achievers.
first_two = high_achievers[:2]

# Step 4: Display both the complete filtered list and its first two items.
print("High achievers:", high_achievers)
print("First two:", first_two)
````

## Quick Recap

- A **dictionary** connects unique keys to values.
- Square brackets access a known key; `get()` safely handles a possibly missing key.
- A **list comprehension** creates a list with one concise expression.
- An optional `if` condition filters items in a comprehension.
- A **slice** uses `[start:stop:step]`, and the stop position is excluded.
